```python
from VesselProcessAPI import MatchTicket, Present, Tool
import pandas as pd

# 测试 get_quantity_statistics 的 start/end 参数
present = Present()

# 1. 用 start/end
df1, m1, w1 = present.get_quantity_statistics(
    start='2026-09-01', end='2026-09-30',
    expected_shipping_days=45, sheet_name='巴西'
)
print(f'=== start/end 模式 ===')
print(f'行数: {len(df1)}')
print(f'年份: {list(m1.keys())}')
print(m1[2026])
print()

# 2. 用 years
df2, m2, w2 = present.get_quantity_statistics(
    years=2026,
    expected_shipping_days=45, sheet_name='巴西'
)
print(f'=== years 模式 ===')
print(f'行数: {len(df2)}')
print(f'年份: {list(m2.keys())}')
print(m2[2026])
print()

# 3. 同时传 start/end 和 years，start/end 优先
df3, m3, w3 = present.get_quantity_statistics(
    years=2026,
    start='2026-09-01', end='2026-09-30',
    expected_shipping_days=45, sheet_name='巴西'
)
print(f'=== start/end + years (start/end 优先) ===')
print(f'行数: {len(df3)}')
print(f'年份: {list(m3.keys())}')
print(m3[2026])
```
